```python
# cloud masking
from rasterio.warp import Resampling, reproject
import numpy as np
```

```python
item=signed_items[1]
red_url = item.assets['B04'].href
nir_url = item.assets['B08'].href
scl_url = item.assets['SCL'].href

with rasterio.open(red_url) as red_src:
        # 1. Reproject AOI to raster CRS
    gdf_projected = gdf.to_crs(red_src.crs)
    # 2. Get AOI bounds
    minx, miny, maxx, maxy = gdf_projected.geometry.iloc[0].bounds
    # 3. Convert geographic bounds → raster window
    window = from_bounds(
        minx, miny, maxx, maxy,
        transform=red_src.transform
    )

    red_band = red_src.read(1, window=window)
    dst_transform = red_src.window_transform(window)
    dst_shape = red_band.shape
    dst_crs = red_src.crs

with rasterio.open(scl_url) as scl_src:
    window = from_bounds(
        minx, miny, maxx, maxy,
        transform=scl_src.transform
    )
    scl_band = scl_src.read(1, window=window)
    source_transform = scl_src.window_transform(window=window)
    source_crs = scl_src.crs
    source_shape = scl_band.shape
    scl_dtype = scl_src.dtypes

empty_array = np.empty(dst_shape, dtype=scl_dtype[0])
reproject(source=scl_band,
          destination=empty_array,
          src_transform=source_transform,
          src_crs=source_crs,
          dst_transform=dst_transform,
          dst_crs=dst_crs,
          resampling=Resampling.nearest)

valid_pixel = ~np.isin(empty_array, [3, 8, 9, 10, 11])
masked_red_band = np.where(valid_pixel, red_band, np.nan)
masked_red_band = masked_red_band*0.0001

```
