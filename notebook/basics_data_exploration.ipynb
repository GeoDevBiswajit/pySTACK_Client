```python
# set pystac_client logger to DEBUG to see API calls
import logging

from pystac_client import Client

logging.basicConfig()
logger = logging.getLogger("pystac_client")
logger.setLevel(logging.DEBUG)
```

```python
URL = "https://planetarycomputer.microsoft.com/api/stac/v1"

# custom headers
headers = []

cat = Client.open(URL, headers=headers)
cat
```

### CollectionClient

```python
# get_collections function will iterate through the Collections in the Catalog
# Check the available products
for collection in cat.get_collections():
    print(collection)
```

```python
collection_search  = cat.collection_search(q='ASTER')
for result in collection_search.collections():
    print(result.id, result.description, sep='\n')
```
